```python
# Reimportar las librerías necesarias y volver a crear el DataFrame
import numpy as np
import pandas as pd

# Parámetros de la simulación
lam = 5  # Tasa promedio de llegadas por hora (puedes ajustar este valor)
horas = ["{:02d}:00".format(h) for h in range(8, 20)]  # Desde 8am hasta 8pm (12 horas)
dias_semana = ["Lunes", "Martes", "Miércoles", "Jueves", "Viernes"]

# Simulación: número de clientes por hora y por día
clientes = np.random.poisson(lam, size=(len(horas), len(dias_semana)))

# Crear un DataFrame con los resultados
df_clientes = pd.DataFrame(clientes, index=horas, columns=dias_semana)

# Guardar el archivo en formato Excel
file_path = "simulacion_clientes_poisson.xlsx"
df_clientes.to_excel(file_path)

file_path
```

```python
import numpy as np
import pandas as pd
from datetime import datetime, timedelta

# Parámetros de la simulación
llegadas_por_hora = 3  # Tasa promedio de llegadas por hora (lambda para la distribución exponencial)
num_pacientes = 50  # Número de pacientes a simular

# Simulación del tiempo entre llegadas usando la distribución exponencial
tiempo_entre_llegadas = np.random.exponential(scale=1/llegadas_por_hora, size=num_pacientes)

# Simulación de las horas de llegada acumuladas
hora_inicio = datetime(2024, 9, 12, 8, 0)  # Fecha y hora de inicio: 8:00 AM del 12 de septiembre de 2024
horas_llegadas = [hora_inicio + timedelta(hours=sum(tiempo_entre_llegadas[:i])) for i in range(1, num_pacientes + 1)]

# Crear un DataFrame con los resultados
fechas = [hora.date() for hora in horas_llegadas]
horas = [hora.time() for hora in horas_llegadas]
df_llegadas = pd.DataFrame({
    "Fecha": fechas,
    "Hora de llegada": horas,
    "Tiempo entre llegadas (horas)": tiempo_entre_llegadas,
    "Número de pacientes": range(1, num_pacientes + 1)
})

# Guardar el archivo en formato Excel
file_path = "simulacion_tiempo_entre_llegadas_exponencial.xlsx"
df_llegadas.to_excel(file_path, index=False)

file_path
```
