```python
def biseccion(f, a, b, tolerancia):
    """
    Método de bisección para encontrar una raíz de la función f en el intervalo [a, b].

    Parámetros:
    f: función
    a: límite inferior del intervalo
    b: límite superior del intervalo
    tolerancia: criterio de parada

    Retorna:
    La aproximación de la raíz de f en el intervalo [a, b].
    """

    # Verificar que la raíz existe en el intervalo dado
    if f(a) * f(b) > 0:
        print("La función debe tener signos opuestos en a y b.")
        return None

    # Iterar hasta que el intervalo sea lo suficientemente pequeño
    while (b - a) / 2.0 > tolerancia:
        c = (a + b) / 2.0  # Punto medio
        if f(c) == 0:
            return c  # c es la raíz exacta
        elif f(a) * f(c) < 0:
            b = c  # La raíz está en [a, c]
        else:
            a = c  # La raíz está en [c, b]

    return (a + b) / 2.0  # Retornar la aproximación de la raíz

# Ejemplo de uso
# Definir la función de la cual queremos encontrar la raíz
def f(x):
    return x**2 - 4

# Definir el intervalo [a, b] y la tolerancia
a = 1
b = 3
tolerancia = 0.000001

# Encontrar la raíz usando el método de bisección
raiz = biseccion(f, a, b, tolerancia)

# Imprimir el resultado
if raiz is not None:
    print(f"La raíz aproximada es: {raiz}")
```

```python
%%R
f <- function(x) {
  sqrt(x) - cos(x)
}

biseccion <- function(f, a, b, tolerancia) {
  if (f(a) * f(b) > 0) {
    print("La función debe tener signos opuestos en a y b.")
    return(NULL)
  }

  while ((b - a) / 2 > tolerancia) {
    c <- (a + b) / 2
    if (f(c) == 0) {
      return(c)
    } else if (f(a) * f(c) < 0) {
      b <- c
    } else {
      a <- c
    }
  }

  return((a + b) / 2)
}

# Definir el intervalo [a, b] y la tolerancia
a <- 0
b <- 1
tolerancia <- 0.0001

# Encontrar la raíz usando el método de bisección
raiz <- biseccion(f, a, b, tolerancia)

# Imprimir el resultado
if (!is.null(raiz)) {
  print(paste("La raíz aproximada es:", raiz))
}
```
