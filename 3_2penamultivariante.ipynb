```python
import numpy as np

# Datos
X1 = np.array([2, 1, 2, 2])
X2 = np.array([3, 5, 2, 3])
X3 = np.array([-1, -2, 1, 1])
datos = np.vstack((X1, X2, X3)).T

# Vector de medias
medias = np.mean(datos, axis=0)
print("Vector de medias:")
print(medias)

# Matriz de varianzas y covarianzas
S = np.cov(datos, rowvar=False)
print("Matriz de varianzas y covarianzas:")
print(S)

# Varianza generalizada (determinante de S)
var_generalizada = np.linalg.det(S)
print("Varianza generalizada:")
print(var_generalizada)

# Matriz de correlación
R = np.corrcoef(datos, rowvar=False)
print("Matriz de correlación:")
print(R)

# Raíz y vector característico mayor
valores_propios, vectores_propios = np.linalg.eig(S)
valor_propio_mayor = np.max(valores_propios)
vector_propio_mayor = vectores_propios[:, np.argmax(valores_propios)]
print("Valor propio mayor:")
print(valor_propio_mayor)
print("Vector propio asociado:")
print(vector_propio_mayor)
```
