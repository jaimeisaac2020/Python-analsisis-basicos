```python
import yfinance as yf
import pandas as pd

# --- 1. Definir los Tickers y el Rango de Fechas ---

# Tickers para Google (Alphabet Inc. Class A) y Amazon
tickers = ['GOOGL', 'AMZN']

# Rango de fechas.
# Como 2025 es futuro, yfinance descargará datos hasta la fecha más reciente disponible.
# Si ejecutas esto en 2026, obtendrá todos los datos de 2025.
start_date = '2025-01-01'
end_date = '2025-12-31'

print(f"Intentando descargar datos para {', '.join(tickers)} desde {start_date} hasta {end_date}...")

# --- 2. Descargar los Datos ---

try:
    # yf.download() descarga los datos OHLC (Open, High, Low, Close), Volumen, etc.
    # Podemos especificar que solo queremos el precio de cierre ('Adj Close' es el recomendado
    # porque ajusta por dividendos y splits).
    data = yf.download(tickers, start=start_date, end=end_date)

    # Verificar si se descargaron datos
    if data.empty:
        print("\nNo se encontraron datos para el período especificado.")
        print("Esto es esperado si la fecha de inicio es en el futuro.")
        print("El script funcionará correctamente una vez que existan datos para 2025.")
    else:
        # Seleccionar solo los precios de cierre ajustados
        closing_prices = data['Adj Close']

        # Renombrar las columnas para mayor claridad
        closing_prices = closing_prices.rename(columns={'GOOGL': 'GOOGL_Close', 'AMZN': 'AMZN_Close'})

        # --- 3. Mostrar y Guardar los Datos ---

        print("\n--- Primeras 5 filas de datos descargados ---")
        print(closing_prices.head())

        print("\n--- Últimas 5 filas de datos descargados ---")
        print(closing_prices.tail())

        # Guardar los datos en un archivo CSV
        file_name = 'precios_2025_google_amazon.csv'
        closing_prices.to_csv(file_name)

        print(f"\n¡Éxito! Los datos han sido guardados en el archivo '{file_name}'")

except Exception as e:
    print(f"\nOcurrió un error durante la descarga: {e}")
    print("Verifica tu conexión a internet o si los tickers son correctos.")
```

```python
import yfinance as yf
import pandas as pd

# --- 1. Definir los Tickers y el Rango de Fechas ---

tickers = ['GOOGL', 'AMZN']
start_date = '2025-01-01'
end_date = '2025-12-31'

print(f"Descargando precios de cierre para {', '.join(tickers)} desde {start_date} hasta {end_date}...")

# --- 2. Descargar solo los Precios de Cierre Ajustados ---

try:
    # El parámetro 'auto_adjust=True' le dice a yfinance que trabaje directamente
    # con los precios ajustados. Esto simplifica el DataFrame resultante.
    # El parámetro 'progress=False' oculta la barra de progreso de la descarga.
    data = yf.download(tickers, start=start_date, end=end_date, progress=False)

    # Cuando se descargan múltiples tickers, yfinance crea un MultiIndex en las columnas.
    # Seleccionamos solo la columna 'Close' que, con auto_adjust=True, ya está ajustada.
    closing_prices = data['Close']

    # Verificar si se descargaron datos
    if closing_prices.empty:
        print("\nNo se encontraron datos para el período especificado.")
        print("Esto es normal si la fecha de inicio es en el futuro.")
    else:
        # Renombrar las columnas para que coincidan con tu dataset
        closing_prices = closing_prices.rename(columns={'GOOGL': 'GOOGL_Close', 'AMZN': 'AMZN_Close'})

        # --- 3. Mostrar y Guardar los Datos ---

        print("\n--- Primeras 5 filas de precios de cierre ---")
        print(closing_prices.head())

        print("\n--- Últimas 5 filas de precios de cierre ---")
        print(closing_prices.tail())

        # Guardar los datos en un archivo CSV
        file_name = 'precios_cierre_2025_google_amazon.csv'
        closing_prices.to_csv(file_name)

        print(f"\n¡Éxito! Los datos de cierre han sido guardados en '{file_name}'")

except Exception as e:
    print(f"\nOcurrió un error: {e}")
```
