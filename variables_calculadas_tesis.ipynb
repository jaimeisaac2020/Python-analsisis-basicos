```python
import pandas as pd
import numpy as np

# --- 1. Carga y Preparación de Datos ---

# Cargar el conjunto de datos
try:
    df = pd.read_csv('dataset_completo_google_amazon.csv')
    print("Archivo CSV cargado exitosamente.")
except FileNotFoundError:
    print("Error: El archivo 'dataset_completo_google_amazon.csv' no se encontró.")
    print("Asegúrate de que el archivo esté en la misma carpeta que este script.")
    exit()

# Convertir la columna 'Date' a formato datetime y ordenar
df['Date'] = pd.to_datetime(df['Date'])
df = df.sort_values(by='Date').reset_index(drop=True)

print("\nPrimeras 5 filas del dataset original:")
print(df.head())

# --- 2. Cálculo de los Retornos Diarios (Return_1d) ---

print("\nCalculando los retornos diarios...")
df['GOOGL_Return_1d_Calculado'] = df['GOOGL_Close'].pct_change(1)
df['AMZN_Return_1d_Calculado'] = df['AMZN_Close'].pct_change(1)
print("Cálculo de retornos completado.")

# --- 3. Cálculo de la Volatilidad a 20 días (Volatility_20) ---

print("\nCalculando la volatilidad de 20 días...")
window_size = 20
df['GOOGL_Volatility_20_Calculada'] = df['GOOGL_Close'].rolling(window=window_size).std()
df['AMZN_Volatility_20_Calculada'] = df['AMZN_Close'].rolling(window=window_size).std()
print("Cálculo de volatilidad completado.")

# --- 4. Verificación de los Resultados ---

print("\n--- Verificación de Resultados ---")
columns_to_compare = [
    'Date', 'GOOGL_Return_1d', 'GOOGL_Return_1d_Calculado',
    'AMZN_Return_1d', 'AMZN_Return_1d_Calculado',
    'GOOGL_Volatility_20', 'GOOGL_Volatility_20_Calculada',
    'AMZN_Volatility_20', 'AMZN_Volatility_20_Calculada'
]

first_valid_index = df['GOOGL_Volatility_20_Calculada'].first_valid_index()
if first_valid_index is not None:
    print("\nDatos a partir de la primera fecha con cálculo de volatilidad válido:")
    print(df.loc[first_valid_index-2 : first_valid_index+5, columns_to_compare].to_string())

# --- Verificación final CORREGIDA ---
# Comparamos a partir de la segunda fila para los retornos,
# y a partir de la fila 20 (índice 19) para la volatilidad para alinear los datos.

return_diff_googl = np.allclose(df['GOOGL_Return_1d'].iloc[1:], df['GOOGL_Return_1d_Calculado'].iloc[1:]) # <-- CAMBIO AQUÍ
vol_diff_googl = np.allclose(df['GOOGL_Volatility_20'].iloc[19:], df['GOOGL_Volatility_20_Calculada'].iloc[19:]) # <-- CAMBIO AQUÍ

return_diff_amzn = np.allclose(df['AMZN_Return_1d'].iloc[1:], df['AMZN_Return_1d_Calculado'].iloc[1:]) # <-- CAMBIO AQUÍ
vol_diff_amzn = np.allclose(df['AMZN_Volatility_20'].iloc[19:], df['AMZN_Volatility_20_Calculada'].iloc[19:]) # <-- CAMBIO AQUÍ

print("\n--- Resumen de la Verificación ---")
print(f"¿Los retornos calculados de GOOGL coinciden con los originales? -> {return_diff_googl}")
print(f"¿La volatilidad calculada de GOOGL coincide con la original? -> {vol_diff_googl}")
print(f"¿Los retornos calculados de AMZN coinciden con los originales? -> {return_diff_amzn}")
print(f"¿La volatilidad calculada de AMZN coincide con la original? -> {vol_diff_amzn}")

if all([return_diff_googl, vol_diff_googl, return_diff_amzn, vol_diff_amzn]):
    print("\n¡Éxito! Todos los cálculos han sido replicados correctamente.")
else:
    print("\nAtención: Algunas columnas calculadas no coinciden exactamente con las originales. Revisa las diferencias.")
```
